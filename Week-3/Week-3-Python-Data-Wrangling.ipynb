## 1. Load the Dataset


```python
import pandas as pd

df = pd.read_csv("data.csv")
df.head()
```

## 2. Check Dataset Information

```python
df.info()
```

## 3. Check Missing Values

```python
df.isnull().sum()
```

## 4. Handle Missing Values

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
df["Date"] = df["Date"].fillna("Unknown")

df.isnull().sum()
```

## 5. Filter Rows

```python
filtered_df = df[df["Calories"] > 400]

filtered_df
```

## 6. Create a New Column

```python
df["Calories_per_Duration"] = df["Calories"] / df["Duration"]

df.head()
```

## 7. Results

- Missing values were handled successfully.
- Rows with Calories greater than 400 were filtered.
- A new `Calories_per_Duration` column was created.
