# Week 2 – Python for Data Analysis
## 1. Load a CSV File and Display Basic Information


```python
import pandas as pd
```

```python
data = """Product,Category,Revenue
Laptop,Technology,50000
Chair,Furniture,15000
Notebook,Office Supplies,5000
Phone,Technology,30000
Table,Furniture,20000
"""

with open("sales_data.csv", "w") as file:
    file.write(data)
```

```python
df = pd.read_csv("sales_data.csv")
df.info()
```

### Result
The CSV file was successfully loaded using Pandas, and basic information about the dataset was displayed using `df.info()`.

## 2. Handle Missing Values and Duplicates

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
df = df.dropna()

print("Missing values after handling:")
print(df.isnull().sum())

print("\nDuplicate rows after handling:")
print(df.duplicated().sum())
```

### Result
Missing values and duplicate rows were checked and handled using Pandas. The dataset contains no missing values or duplicate rows.

## 3. Group Data by Category and Find Total Revenue

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

print(category_revenue)
```

### Result
The data was grouped by category and total revenue was calculated using `groupby()` and `sum()`.

## 4. Sort Data by Multiple Columns

```python
sorted_df = df.sort_values(
    by=["Category", "Revenue"],
    ascending=[True, False]
)

print(sorted_df)
```

### Result
The data was sorted by Category in ascending order and Revenue in descending order using Pandas `sort_values()`.

## 5. Create a Correlation Matrix


```python
correlation_matrix = df.corr(numeric_only=True)

print(correlation_matrix)
```

### Result
A correlation matrix was created for the numerical columns using Pandas. Since the dataset contains only one numerical column, Revenue has a correlation of 1.0 with itself.
